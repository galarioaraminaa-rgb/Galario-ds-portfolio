**Instruction:** Perform a single forward pass through the network and compute the error.

Given: input $x = [1, 0, 1]^T$, target $y = 1$, and the ReLU activation $f(z) = \max(0, z)$.

## Given Parameters

Hidden layer weights: $w_{11}=0.2,\ w_{12}=-0.3,\ w_{13}=0.4,\ w_{14}=0.1,\ w_{15}=-0.5,\ w_{16}=0.2$

Output layer weights: $w_{21}=-0.3,\ w_{22}=-0.2$

Biases: $\theta_1=-0.4,\ \theta_2=0.2,\ \theta_3=0.1$

## Step 1 — Hidden Layer

$Z_1 = (0.2)(1) + (0.4)(0) + (-0.5)(1) + (-0.4) = -0.7$, so $H_1 = \max(0, -0.7) = 0$

$Z_2 = (-0.3)(1) + (0.1)(0) + (0.2)(1) + 0.2 = 0.1$, so $H_2 = \max(0, 0.1) = 0.1$

## Step 2 — Output Layer

$Z_3 = (-0.3)(0) + (-0.2)(0.1) + 0.1 = 0.08$

$\hat{y} = \max(0, 0.08) = 0.08$

## Step 3 — Error

Using the lesson's convention, $\delta = \hat{y} - y = 0.08 - 1 = \mathbf{-0.92}$

For reference, the squared error is $(1 - 0.08)^2 = 0.8464$.

## Verify with Python

```python
# input and target
x1, x2, x3 = 1, 0, 1
y = 1

# ReLU
def relu(z):
    return max(0, z)

# hidden layer
z1 = 0.2*x1 + 0.4*x2 + (-0.5)*x3 + (-0.4)
z2 = -0.3*x1 + 0.1*x2 + 0.2*x3 + 0.2
h1 = relu(z1)
h2 = relu(z2)

# output layer
z3 = -0.3*h1 + (-0.2)*h2 + 0.1
y_hat = relu(z3)

# error
error = y_hat - y

print("Z1 =", round(z1, 4), " H1 =", round(h1, 4))
print("Z2 =", round(z2, 4), " H2 =", round(h2, 4))
print("Z3 =", round(z3, 4))
print("Prediction =", round(y_hat, 4))
print("Error (y_hat - y) =", round(error, 4))
print("Squared error =", round(error**2, 4))
```
