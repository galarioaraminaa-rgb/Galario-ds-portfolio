## 1. Standard Imports

```python
import numpy as np
import torch

print("NumPy:", np.__version__)
print("PyTorch:", torch.__version__)
```

## 2. Create `set_seed(seed)`

The function seeds both NumPy and PyTorch and returns nothing.

```python
def set_seed(seed):
    np.random.seed(seed)
    torch.manual_seed(seed)
```

## 3. Create NumPy Array `arr`

Six random integers from 0 (inclusive) to 5 (exclusive), using seed 42. The `int32` dtype lets us convert to `int64` in Task 5.

```python
set_seed(42)

arr = np.random.randint(0, 5, 6).astype("int32")
print("arr =", arr)
print("dtype =", arr.dtype)
```

## 4. Create Tensor `x` from `arr`

```python
x = torch.from_numpy(arr)
print(x)
print("dtype =", x.dtype)
```

## 5. Change `x` from `int32` to `int64`

```python
x = x.to(torch.int64)
print(x)
print("dtype =", x.dtype)
```

## 6. Reshape `x` into a 3 × 2 Tensor

```python
x = x.reshape(3, 2)
print(x)
print("shape =", x.shape)
```

## 7. Return the Right-Hand Column of `x`

```python
print(x[:, 1])
```

## 8. Return the Squares of `x` Without Changing `x`

`x ** 2` creates a new tensor, so `x` stays the same.

```python
print("Squared:")
print(x ** 2)

print("\nx is unchanged:")
print(x)
```

## 9. Create Tensor `y`

`x` has shape (3, 2), so for matrix multiplication `y` needs 2 rows. With the same 6 elements as `x`, its shape is (2, 3).

```python
torch.manual_seed(42)

y = torch.randint(0, 5, (2, 3))
print(y)
print("shape =", y.shape)
print("elements =", y.numel())
```

## 10. Matrix Product of `x` and `y`

(3, 2) @ (2, 3) gives a (3, 3) result.

```python
product = x @ y
print(product)
print("shape =", product.shape)
```
