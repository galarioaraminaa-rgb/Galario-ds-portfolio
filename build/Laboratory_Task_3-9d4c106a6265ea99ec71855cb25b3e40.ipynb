## Setup

Row *i* of `W1` holds the weights going into hidden neuron *i*, so `W1` has shape (2, 3).

```python
import numpy as np

x = np.array([1, 0, 1])
y = 1
lr = 0.001

W1 = np.array([[0.2, 0.4, -0.5],
               [-0.3, 0.1, 0.2]])
b1 = np.array([-0.4, 0.2])

W2 = np.array([-0.3, -0.2])
b2 = 0.1

def relu(z):
    return np.maximum(0, z)
```

## 1. Forward Propagation

```python
z1 = W1 @ x + b1
h = relu(z1)
z2 = W2 @ h + b2
y_hat = relu(z2)

error = y_hat - y
loss = 0.5 * error**2

print("Hidden z:", z1)
print("Hidden h:", h)
print("Output z:", z2)
print("Prediction:", y_hat)
print("Error:", error)
print("Loss:", loss)
```

## 2. Backward Propagation

For $E = \frac{1}{2}(\hat{y} - y)^2$, the output delta is $(\hat{y} - y) \cdot f'(Z_3)$. The delta is then passed back to the hidden layer using the output weights and the ReLU derivative ($1$ if $z > 0$, otherwise $0$).

```python
# output layer
delta_out = error * (z2 > 0)
dW2 = delta_out * h
db2 = delta_out

# hidden layer
delta_hidden = delta_out * W2 * (z1 > 0)
dW1 = np.outer(delta_hidden, x)
db1 = delta_hidden

print("Output delta:", delta_out)
print("Hidden delta:", delta_hidden)
print("dW2:", dW2)
print("db2:", db2)
print("dW1:\n", dW1)
print("db1:", db1)
```

## 3. Update Weights and Biases

$w_{new} = w - \eta \cdot \frac{\partial E}{\partial w}$

```python
W1_new = W1 - lr * dW1
b1_new = b1 - lr * db1
W2_new = W2 - lr * dW2
b2_new = b2 - lr * db2

print("New W1:\n", W1_new)
print("New b1:", b1_new)
print("New W2:", W2_new)
print("New b2:", b2_new)
```

## 4. Check the Result After One Update

```python
h_new = relu(W1_new @ x + b1_new)
y_hat_new = relu(W2_new @ h_new + b2_new)
loss_new = 0.5 * (y_hat_new - y)**2

print("New prediction:", y_hat_new)
print("New loss:", loss_new)
print("Loss decreased:", loss_new < loss)
```
